```python
import numpy as np
```

```python
a=np.arange(5,101,5)
b=a.reshape(4,5)
c=np.ndim(b)
d=np.shape(b)
print("1D array: \n",a)
print("Reshaped array: \n",b)
print("The number of dimensions: ",c)
print("The shape of the matrix: ",d)
print("The total number of elements: ",b.size)
print("The data type of the elements: ",b.dtype)
```

```python
prices = np.array([120, 250, 80, 450, 300]) 
quantities = np.array([4, 2, 10, 1, 3])

revenue = prices*quantities
total_revenue=np.sum(revenue)
average_revenue = np.mean(revenue)

print("Revenue generated by each individual product: ",revenue)
print("Overall combined revenue across all 5 products: ",total_revenue)
print("Average revenue per product: ",average_revenue)
print("Index position of the highest-earning product: ",np.argmax(revenue))
```

```python
rainfall = np.array([12.5, -1.0, 0.0, 18.2, -99.0, 5.4, 22.1, 0.0, -1.0, 14.8])
rainfall[rainfall<0]=0
print("Valid rainfall readings: ",rainfall[rainfall>0])
print("Cleaned array: ",rainfall)
print("Days recorded heavy rainfall: ",np.sum(rainfall>10))
```

```python
np.random.seed(10)

scores = np.random.randint(20,100,12)
status = np.where(scores>=40, "Pass","Fail")
updated_scores = np.where(scores<40, scores+5, scores)

print("Scores: ",scores)
print("Status: ",status)
print("Updated scores: ",updated_scores)
```

```python
np.random.seed(10)

marks = np.random.randint(50,100,(6,4))
average = np.mean(marks, axis=1)
highest = np.max(marks, axis=0)

print("Marks: ",marks)
print("Average: ",np.round(average,2))
print("Highest mark scored in each subject: ",highest)
print("Rows belonging to students whose average mark is strictly greater than 75: \n",marks[average>75])
```

```python
houses = np.array([ [1200, 2, 55], 
                    [1800, 3, 85], 
                    [950,  1, 40], 
                    [2400, 4, 120], 
                    [1500, 3, 70] ])

Xmin=np.min(houses, axis=0)
Xmax=np.max(houses, axis=0)
Xnorm = (houses - Xmin) / (Xmax - Xmin)

print("Normalized array: \n",np.round(Xnorm,3))
```

```python
morning_sales = np.array([ [15, 20], [10, 12], [25, 30] ]) 
evening_sales = np.array([ [18, 22], [14, 15], [20, 28] ])

fullday_sales = np.concatenate((morning_sales, evening_sales), axis=1)
branch = np.array([[12, 16, 15, 19]])
final= np.concatenate((fullday_sales,branch), axis=0)

print("Full-day sales: \n", fullday_sales.reshape(3,4))
print("New matrix: \n",final)
print("Total units sold in each: ",np.sum(final, axis=0))
```

```python
ratings = np.array([5, 4, 3, 5, 2, 5, 4, 1, 3, 5, 4, 4, 2, 5, 3, 4, 5, 1, 4, 3])

unique_rating, counts = np.unique(ratings, return_counts=True)
percentages = (counts / len(ratings)) * 100

for rating, count, percentage in zip(unique_rating, counts, percentages):
    print(f"{rating}: {count} responses ({percentage:.1f}%)")
```

```python
emp_ids = np.array(["E101", "E102", "E103", "E104", "E105", "E106"]) 
scores  = np.array([78, 92, 65, 88, 95, 71])

sorted_indices = np.argsort(scores)[::-1]
sorted_emp_ids = emp_ids[sorted_indices]
sorted_scores = scores[sorted_indices]

print("Top 3 Employees:")
for i in range(3):
    print(f"{sorted_emp_ids[i]}: {sorted_scores[i]}")
```
