STACK METHODS

stack
hstack
vstack

```python
j=np.array([1,2,3,4])
k=np.array([5,6,7,8])
l=np.stack((j,k),axis=1)
print(l)
```

```python
print(np.hstack((j,k)))
print(np.vstack((j,k)))
```

```python
a=[1,2,3,4]
b=a.copy()
print(a==b)
a[0]=10
print(a)
print(b)

a=np.array(a)
b=a.view()
a[1]=20
print(a)
print(b)
```
